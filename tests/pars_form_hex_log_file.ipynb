# Pars form hex log file

## 1. Probedaten einlesen 

```python
from pathlib import Path

# Pfad festlegen
project_path = Path.cwd().parent
folder = 'example_output'
filename = 'hex_24102025_1434'
filepath = project_path / folder / filename

# Datei einlesen
try:
    with open(filepath, 'r') as f:
        file = f.read()
except Exception as e:
    print(f"Fehler beim Lesen der Datei: {e}")

# Checken
print(f"File output: {file[:100]}")
print(type(file))
```

## 2. Sting in Bin umwandeln 

```python
import ast

# Umwandlung von String zu Binärdaten
try:
    # bytes-Objekt erstellen 
    binary_data = ast.literal_eval(file)
    
    print(f"File output: {binary_data[:50]}")
    print(type(binary_data))
    
except (ValueError, SyntaxError) as e:
    print(f"Fehler beim Umwandeln des Strings in Binärdaten: {e}")
```

## 3.1 Umwandlung in eine Liste von Integern: 

```python
integer_liste = list(binary_data)

print(f"Int list: {integer_liste[:30]}")
print(type(integer_liste))
print(type(integer_liste[0]))
```

## 3.2 Umwandlen in binär

```python
# Bit String
bit_string = "".join(format(byte, '08b') for byte in binary_data)
print(f"Bit string: {bit_string[:240]}")
print(type(bit_string))

# Byte Liste
bit_string_liste = [format(byte, '08b') for byte in binary_data]
print(f"Bit string list: {bit_string_liste[:20]}")
print(type(bit_string_liste))
print(type(bit_string_liste[0]))
```

## 3.3 Umwandeln in hex list 

```python
hex_liste = [format(byte, '02x') for byte in binary_data]

print(f"Hex list: {hex_liste[:30]}")   
print(type(hex_liste))
print(type(hex_liste[0])) 
```

## 4. Vergleich 

```python
print(f"File output: {binary_data[:50]}")
print(f"Hex list: {hex_liste[:14]}")
print(f"Int list: {integer_liste[:14]}") 
print(f"Bit string list: {bit_string_liste[:14]}")
```

## Erklärung der Datenstruktur (\x00 + Buchstaben )

  Schauen wir uns das erste Stück an: b'\x00d'

   * Dies sind zwei Bytes (=2*8bit=16bit): Das erste ist \x00 und das zweite ist d.
   * In Ihrer Int list sehen Sie das als [0, 100].
   * Was ist `\x00`? Das ist das erste Byte des 2-Byte-Wertes. Man nennt es auch das "Most Significant Byte" (MSB). Sein Wert ist 0.
   * Was sind die Buchstaben (`d`, `z`, `x`...)? Das sind Bytes, deren Zahlenwert im druckbaren ASCII-Bereich liegt. Python ist hier hilfreich und zeigt
     Ihnen direkt das Zeichen anstatt des Hex-Codes. d ist einfach eine andere Schreibweise für \x64, was der Zahl 100 entspricht.
   * Wie wird daraus eine Zahl? Die beiden Bytes \x00 und d (also 0 und 100) bilden zusammen die 16-Bit-Zahl 100. Die Formel dafür lautet: (Erstes Byte *
     256) + Zweites Byte. Also (0 * 256) + 100 = 100.

  Schauen wir uns ein anderes Beispiel aus Ihren Daten an: b'\x01\xd2'
   * Erstes Byte (\x01) = 1
   * Zweites Byte (\xd2) = 210
   * Zusammengesetzter Wert = (1 * 256) + 210 = 466.

  Zusammenfassend zur Datenstruktur:
  Die \x00 ist meistens das erste von zwei Bytes und hat den Wert 0, weil die dargestellte Zahl klein ist (kleiner als 256). Die Buchstaben sind einfach
   eine lesbare Darstellung für das zweite Byte. Die Fragezeichen (?) oder Punkte (.), die Sie in früheren Ausgaben gesehen haben, waren nur Platzhalter
   von meinem Beispielcode für Bytes, die keinem druckbaren Zeichen entsprechen.

## 5. Interpretieren 

```python
import struct
 
def parse_header(data_bytes):
    """
    Parst den Header aus einem bytes-Objekt gemäß der definierten Struktur.
    """
    # Wir prüfen, ob überhaupt genügend Daten für den Header vorhanden sind (8 + 2 + 4 = 14 Bytes)
    if len(data_bytes) < 14:
        print("Fehler: Nicht genügend Daten für den Header vorhanden.")
        return None

    # Magic (erste 8 Bytes) in char
    magic_bytes = data_bytes[0:8]
    magic_string = magic_bytes.decode('ascii', errors='ignore')

    # Packet ID (nächste 2 Bytes) in char
    packetid_bytes = data_bytes[8:10]
    packetid_string = packetid_bytes.decode('ascii', errors='ignore')

    # Length (nächste 4 Bytes) in uint32
    length_bytes = data_bytes[10:14]
    length_uint32 = int.from_bytes(length_bytes, byteorder='big')

    # Wir sammeln alle Informationen in einem Dictionary für eine saubere Rückgabe
    header_info = {
        "magic": magic_string,
        "packet_id": packetid_string,
        "length": length_uint32
    }

    return header_info
   

header = parse_header(binary_data)

# Schöne Ausgabe der Ergebnisse
if header:
    print("--- Analysierte Header-Informationen ---")
    print(f"Magic:     '{header['magic']}'")
    print(f"Packet ID: '{header['packet_id']}'")
    print(f"Length:    {header['length']}")
```
